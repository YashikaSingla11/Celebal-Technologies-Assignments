Assignment - 1 Basics of Pandas
Dataset -Combined_dataset.csv

1. Import the Library

```python
import pandas as pd
import numpy as np
```

2. Load a CSV dataset into a Pandas DataFrame

```python
df = pd.read_csv('Combined_dataset.csv')
```

3. Explore Dataset

```python
print("Shape")
print(df.shape)

print("\nColumns")
print(df.columns.tolist())

print("\nData Types")
print(df.dtypes)

print("\nHead")
display(df.head())

print("\nTail")
display(df.tail())
```

4. Handle Missing Values

```python
print(df.isnull().sum())
```

5. Remove Missing Values

```python
df.dropna(inplace=True)
```

6. Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

7. Filter Rows

```python
display(df[df['rating'] > 4])
```

8. Select Columns

```python
display(df[['title', 'final_price', 'rating']])
```

9. Derived Column

```python
df['discount_amount'] = df['initial_price'] * (df['discount'] / 100)
```

10. Save Cleaned Dataset

```python
df.to_csv('Cleaned_Combined_dataset.csv', index=False)
```
