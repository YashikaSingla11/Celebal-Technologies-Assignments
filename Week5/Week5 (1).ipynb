```python
#Import Spark Session
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Week5_Data_Processing") \
    .getOrCreate()

print("Spark Session Created")
```

```python
# Create sample dataset
data = [
(1,"2025-01-01","West","Electronics",500,25,"Premium","Delhi","Active","2025-01-01 10:00:00","user1@gmail.com","John",101),
(2,"2025-01-02","East","Clothing",300,22,"Premium","Mumbai",None,"2025-01-02 11:00:00","user2@gmail.com","Alice",102),
(3,"2025-01-03","West","Electronics",700,29,"Basic","Delhi","Inactive","2025-01-03 09:30:00",None,"Bob",101),
(4,"2025-01-04","South","Furniture",None,35,"Premium","Chennai","Active","2025-01-04 08:00:00","user4@gmail.com","",103),
(5,"2025-01-05","West","Clothing",450,19,"Premium","Delhi",None,"2025-01-05 12:00:00","user5@gmail.com","David",102),
(1,"2025-01-01","West","Electronics",500,25,"Premium","Delhi","Active","2025-01-01 10:00:00","user1@gmail.com","John",101)
]

columns = ["user_id","transaction_date","region","product_category",
           "sale_amount","age","subscription","city","status",
           "raw_timestamp","email","username","store_id"]

df = spark.createDataFrame(data,columns)

df.show()
```

```python
df.show()
```

```python
# Remove duplicate rows based on user_id and transaction_date
df_no_duplicates = df.dropDuplicates(["user_id","transaction_date"])
# Show result
df_no_duplicates.show()
```

```python
from pyspark.sql.functions import avg
# Filter rows where region is West
# Group by product category and calculate average sales
west_sales = df.filter(df.region=="West")\
.groupBy("product_category")\
.agg(avg("sale_amount").alias("Average_Sales"))
# Display result
west_sales.show()
```

```python
df.na.drop().show()
```

```python
# Replace null values in status column with Unknown
df_status = df.na.fill({"status":"Unknown"})
df_status.show()
```

```python
from pyspark.sql.functions import count

city_count = df.groupBy("city")\
.agg(count("*").alias("Count"))\
.filter("Count > 100")

city_count.show()
```

```python
new_df = df.drop("status")

new_df = new_df.withColumnRenamed(
    "sale_amount",
    "price"
)
```

```python
premium_users = df.filter(
    (df.age>=18) &
    (df.age<=30) &
    (df.subscription=="Premium")
)

premium_users.show()
```

```python
clean_df = df.na.fill({"sale_amount":0})
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType
# Convert raw_timestamp column to TimestampType
df_time = df.withColumn(
    "raw_timestamp",
    col("raw_timestamp").cast(TimestampType())
)
# Rename column to event_time
df_time = df_time.withColumnRenamed(
    "raw_timestamp",
    "event_time"
)

df_time.show()
```

```python
df.groupBy("city").count()
```

```python
from pyspark.sql.functions import col

clean_users = df.filter(
    (col("email").isNotNull()) &
    (col("username") != "")
)

clean_users.show()
```

```python
from pyspark.sql.functions import min,max,mean
# Calculate minimum, maximum and average sale amount
df.agg(
    min("sale_amount").alias("Minimum"),
    max("sale_amount").alias("Maximum"),
    mean("sale_amount").alias("Average")
).show()
```

```python
from pyspark.sql.functions import sum

# Step 1: Remove duplicate rows
# Step 2: Replace null sale_amount with 0
# Step 3: Group by store_id
# Step 4: Calculate total revenue

final_df = df.dropDuplicates()\
             .na.fill({"sale_amount":0})\
             .groupBy("store_id")\
             .agg(sum("sale_amount").alias("Total_Revenue"))

final_df.show()
```

```python
from pyspark.sql.functions import count,sum,avg,min,max

df.groupBy("region").agg(
count("*").alias("Total_Records"),
sum("sale_amount").alias("Total_Sales"),
avg("sale_amount").alias("Average_Sales"),
min("sale_amount").alias("Minimum_Sales"),
max("sale_amount").alias("Maximum_Sales")
).show()
```

```python
# Convert Spark DataFrame to Pandas DataFrame
# Save output into CSV file
final_df.toPandas().to_csv("output.csv", index=False)
```
